# **11. Paper guía**

El artículo científico DengueGNN: Graph-based deep learning for modeling disease spread dynamics and prediction, desarrollado por GulMohamed et al. (2026), se consolida como la referencia principal y el pilar fundamental de esta propuesta de investigación debido a su estricta convergencia metodológica con el objetivo de modelar la dispersión del dengue en entornos geográficos complejos e irregulares. La justificación medular para seleccionar este trabajo como paper base radica en que resuelve con rigor matemático el desafío de integrar la naturaleza dinámica de los factores antrópicos y ambientales dentro de una estructura de grafos espacio-temporales, superando las limitaciones de los enfoques estáticos tradicionales. Para los propósitos de esta investigación, el artículo no solo actúa como un precedente teórico, sino como una guía operativa e instrumental explícita que valida el porqué y el para qué de la transición hacia una arquitectura avanzada como la Dynamic Diffusion Convolutional RNN (D-DCRNN). Al demostrar empíricamente sobre el repositorio global OpenDengue que los grafos variables en el tiempo reducen significativamente el error de pronóstico en horizontes críticos de mediano plazo (1 a 4 semanas), este texto provee el sustento analítico indispensable para legitimar la viabilidad y necesidad de desplegar un marco predictivo similar adaptado a la compleja realidad epidemiológica de los departamentos de Colombia.

Bajo esta premisa, el modelo original de GulMohamed et al. (2026) aporta una arquitectura conceptualmente robusta dividida en cuatro componentes interconectados que sirven como base directa para la formulación del diseño D-DCRNN. El sistema original se compone de un módulo de construcción de grafos epidemiológicos dinámicos, un codificador espacial fundamentado en capas de convolución de grafos (GCN), un codificador temporal que emplea celdas LSTM potenciadas con mecanismos de atención adaptativa, y una capa de fusión de características climáticas y demográficas. El supuesto fundamental de esta arquitectura dictamina que la transmisión del dengue no responde a una estructura geográfica inmutable, sino que los patrones de contagio interregional mutan dinámicamente semana a semana, gobernados de forma no lineal por los flujos fluctuantes de la movilidad humana y las variaciones climáticas locales. Con base en este supuesto, los autores calculan y actualizan periódicamente una matriz de adyacencia híbrida que fusiona la contigüidad física con proyecciones de transporte derivadas de un modelo de gravedad poblacional. Esta formulación matemática exacta es el insumo principal que se requiere para heredar y refinar en la propuesta D-DCRNN, permitiendo transicionar de una convolución de grafos estándar a un proceso de difusión convolucional dinámico que capture con mayor precisión el flujo y la dirección de la propagación del vector en el territorio nacional.

La adopción de este estudio como eje conductor se vuelve completamente persuasiva al analizar su relación analítica con los modelos de referencia o benchmarks de la literatura actual, tales como GRU, LSTM, ConvLSTM, Transformer y las variaciones clásicas de STGNN. En términos de similitudes, el modelo propuesto comparte con LSTM y GRU la capacidad intrínseca para codificar la memoria secuencial de los registros históricos de incidencia en el vector de tiempo, y se asemeja a ConvLSTM y a las redes espacio-temporales tradicionales en el intento conjunto de capturar simultáneamente las dimensiones espaciales y temporales en una sola estructura. Sin embargo, las diferencias marcan la superioridad del enfoque del paper base y justifican su elección: mientras que LSTM, GRU y las arquitecturas Transformer clásicas analizan cada unidad territorial de manera aislada o bajo dependencias puramente secuenciales que pierden las correlaciones geográficas explícitas, las redes ConvLSTM están severamente restringidas al procesamiento de datos estructurados en cuadrículas regulares bidimensionales, lo que las inhabilita para representar fielmente las divisiones administrativas asimétricas, heterogéneas e irregulares de los departamentos de Colombia. Finalmente, la diferencia crucial respecto a las STGNN convencionales de la literatura radica en que estas últimas dependen de una matriz de conectividad fija o basada únicamente en distancias euclidianas; en contraste, el modelo guía rompe este paradigma al reformular la matriz de adyacencia de manera dinámica y guiada por variables externas de movilidad y densidad. Esta última distinción es el argumento teórico maestro que se utilizará para fundamentar las ventajas operativas del diseño D-DCRNN, consolidando al artículo de GulMohamed et al. (2026) como el único antecedente capaz de ofrecer el andamiaje metodológico para integrar con éxito datos climáticos y flujos de transporte humano en un modelo predictivo de alta precisión para la salud pública colombiana.
