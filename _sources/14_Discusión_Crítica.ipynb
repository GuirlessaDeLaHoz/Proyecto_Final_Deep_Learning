# **14. Discusión Crítica**

El análisis comparativo de seis arquitecturas espacio-temporales sobre un esquema de
validación cruzada anidada con tres semillas de inicialización, cuatro folds espaciales
y tres años de prueba independientes constituye uno de los marcos experimentales más
rigurosos reportados en la literatura de predicción hidrológica para el territorio
colombiano. La elección de este diseño no es trivial: al separar simultáneamente la
variabilidad espacial —mediante folds geográficamente disjuntos— y la variabilidad
temporal —mediante años de prueba no vistos durante el entrenamiento—, el esquema
permite estimar el error de generalización real del modelo bajo condiciones que simulan
fielmente el despliegue operativo, evitando la sobreestimación del desempeño que
caracteriza a evaluaciones con particiones aleatorias o temporalmente solapadas. Sobre
este marco, el modelo propuesto **ECT_CV** no solo supera a los cinco benchmarks
considerados, sino que lo hace de manera consistente, estadísticamente verificable y
geográficamente distribuida, lo que convierte su superioridad en un resultado
estructural y no en un artefacto de la configuración experimental.

**Superioridad del modelo propuesto.** En términos de precisión global, el ECT_CV registra
el menor RMSE (3.494 ± 0.97), el menor MAE (1.843 ± 0.72), el menor sMAPE (25.207%) y
el menor RMSE_top10 (9.043), esta última métrica especialmente relevante porque evalúa
el error exclusivamente sobre los eventos de mayor magnitud, donde los errores de
predicción tienen consecuencias operativas directas en términos de alertas tempranas y
gestión de riesgos hidrológicos. Más allá de las métricas de error, el ECT_CV es el
único modelo que supera simultáneamente los umbrales de NSE > 0.70 (NSE = 0.716) y
KGE > 0.70 (KGE = 0.701) en el promedio global, valores que en la literatura hidrológica
se consideran indicativos de un buen ajuste del modelo a la dinámica observada y que
ninguna de las arquitecturas benchmark alcanza. Esta doble superación de umbrales no es
menor: el NSE penaliza errores en la magnitud de los picos, mientras que el KGE
descompone el error en correlación, sesgo relativo y variabilidad relativa, de modo que
un KGE > 0.70 implica que el modelo captura adecuadamente las tres dimensiones del
hidrograma simultáneamente. Los cinco modelos benchmark se quedan sistemáticamente
por debajo de estos umbrales, con LSTM_CV alcanzando el segundo lugar en NSE (0.663)
y KGE (0.650), seguido de STT_CV (NSE = 0.648, KGE = 0.571), STGRU_CV (NSE = 0.599,
KGE = 0.540), ConvLSTM_CV (NSE = 0.509, KGE = 0.441) y STGNN_CV (NSE = 0.282,
KGE = 0.282) como el modelo de peor desempeño global de forma contundente.

La ventaja del ECT_CV adquiere mayor significado cuando se analiza en función de la
Pearson_r: con un valor de 0.863, el ECT_CV supera a todos los benchmarks incluyendo
al STT_CV (0.854), que es la segunda arquitectura más alta en esta métrica. Aunque la
diferencia en valor absoluto es pequeña, su consistencia a través de los tres años de
prueba y los cuatro folds espaciales —donde el ECT_CV mantiene Pearson_r > 0.85 incluso
en 2024 y en el Fold 3— indica que su capacidad para reproducir la estructura temporal
del hidrograma es genuinamente superior y no dependiente de condiciones favorables.

**Evidencia estadística de la superioridad.** La aplicación del Bootstrap pareado con
B = 10,000 iteraciones y n = 39,168 observaciones alineadas proporciona el respaldo
estadístico necesario para traducir las diferencias descriptivas en jerarquías
verificables. Los resultados confirman que **el ECT_CV supera estadísticamente a todos
los benchmarks** con p < 0.05 tanto en RMSE como en MAE, con los intervalos de confianza
al 95% completamente alejados de cero en todos los pares que involucran al modelo
propuesto. La brecha más amplia se observa frente al STGNN_CV (Δ RMSE = 0.263,
IC: [0.257, 0.269]), mientras que la más estrecha —aunque igualmente significativa— se
da frente al STT_CV (Δ RMSE = 0.007, p = 0.035), el benchmark más competitivo del
conjunto. Este último resultado es importante: aunque la diferencia numérica entre
ECT_CV y STT_CV es pequeña en RMSE, el Bootstrap la identifica como estadísticamente
real, lo que descarta que la ventaja del modelo propuesto sea ruido muestral. En el
interior del grupo de benchmarks, el único par donde el Bootstrap no permite establecer
jerarquía es STT_CV vs. LSTM_CV (Δ RMSE = 0.0003, p = 0.906), resultado que indica
equivalencia estadística entre ambos modelos en esta métrica y que tiene implicaciones
prácticas directas sobre la selección de arquitectura en contextos con restricciones
computacionales. El ranking estadísticamente respaldado queda entonces como
**ECT_CV > STT_CV ≈ LSTM_CV > STGRU_CV > ConvLSTM_CV > STGNN_CV**, donde el símbolo
≈ refleja la única no-distinción identificada por el Bootstrap en el conjunto de 15 pares
evaluados.

**Generalización espacial y temporal.** Uno de los aspectos más reveladores del análisis
es que la superioridad del ECT_CV no se concentra en condiciones favorables sino que
se distribuye estructuralmente a través de escenarios de creciente dificultad. En el
Fold 3 (región Andina-Pacífica, 12 departamentos), donde todos los modelos alcanzan
su mejor desempeño, el ECT_CV amplifica la ventaja: NSE cercanos a 0.80–0.85 y KGE
sostenidamente por encima de 0.80, con Pearson_r que supera 0.90 en 2023 y 2024.
En el Fold 1 (Costa Caribe), mantiene NSE entre 0.65 y 0.80 con Bias contenido
(|Bias| < 0.5), mientras que benchmarks como ConvLSTM_CV acumulan sesgos superiores
a −2.5 en 2024. En el Fold 2 (región Suroccidental, de alta complejidad orográfica),
el ECT_CV contiene el RMSE_top10 por debajo de 20 cuando los benchmarks lo superan
ampliamente en 2024, y su NSE permanece por encima de 0.45 en ese año, nivel que ningún
otro modelo sostiene consistentemente en ese fold. Y en el Fold 4 (Amazonía, el escenario
más adverso con solo tres departamentos y baja densidad de estaciones), el ECT_CV es el
único modelo que muestra una trayectoria de recuperación entre 2023 y 2024 en NSE y
Skill Score, mientras el resto estanca o deteriora su desempeño, lo que apunta a una
capacidad diferencial de adaptación ante distribuciones de datos extremadamente escasas.
Esta robustez geográfica es consistente con el índice de robustez normalizada más bajo
del conjunto (0.374), que cuantifica que la variabilidad promedio del ECT_CV entre seeds,
folds y años representa apenas el 37% de la variabilidad máxima observada, frente al
90.8% del ConvLSTM_CV, el modelo más inestable.

Temporalmente, el año 2024 actúa como un test de estrés natural para todos los modelos:
los errores escalan en todas las arquitecturas, posiblemente asociados a condiciones
hidrometeorológicas atípicas ligadas a eventos ENSO o anomalías de precipitación no
representadas en los datos de entrenamiento (2015–2021). Bajo este estrés, el ECT_CV
es el modelo que mejor amortigua el deterioro: mientras el RMSE de LSTM_CV escala de
2.55 (2022) a 5.94 (2024) —un incremento del 133%— y el de ConvLSTM_CV pasa de 3.42
a 6.92 —un 102%—, el ECT_CV muestra un deterioro relativo menor, manteniendo métricas
de ajuste global como NSE y Skill Score por encima de los benchmarks en ese año crítico.
Esta resiliencia temporal, combinada con la robustez espacial documentada, constituye
el argumento más sólido a favor del ECT_CV como arquitectura para despliegue operativo:
un modelo que no solo es preciso en condiciones estándar sino que degrada menos bajo
condiciones extremas es, desde la perspectiva de la gestión de riesgos hidrológicos,
sustancialmente más valioso que uno que sea marginalmente más preciso en años normales
pero colapse ante eventos atípicos.

**Trade-off desempeño-costo y recomendaciones de selección.** La superioridad del ECT_CV
conlleva un trade-off real en costo computacional que no puede ignorarse en una discusión
rigurosa. Al incorporar mecanismos de atención espacio-temporal, codificación de contexto
temporal explícita y un mayor número de parámetros entrenables, el ECT_CV demanda
significativamente más recursos que LSTM_CV o STGRU_CV. Sin embargo, la evidencia
disponible indica que esta inversión es estadísticamente justificada: el Bootstrap
confirma que las diferencias en RMSE y MAE no son marginales frente a ningún benchmark,
y el índice de robustez cuantifica que la ganancia en estabilidad es tan importante como
la ganancia en precisión. En contextos operativos con restricciones de infraestructura
severas —sistemas embebidos, reentrenamiento en tiempo real, o redes de monitoreo con
alta frecuencia de actualización—, el LSTM_CV representa la alternativa más eficiente
del grupo de benchmarks: es estadísticamente equivalente al STT_CV en RMSE (p = 0.906),
superior al STGRU_CV y al ConvLSTM_CV con significancia estadística, tiene un índice de
robustez de 0.562 y un gap de generalización de 0.030 ± 0.049, indicativo de sobreajuste
leve pero sistemáticamente controlado. El STT_CV, aunque equivalente al LSTM_CV en RMSE,
presenta el CV más bajo en Pearson_r (1.99%) de todo el conjunto y un desempeño superior
en MAE en el Bootstrap, lo que lo hace preferible cuando la reproducibilidad de la
estructura temporal del hidrograma es prioritaria sobre la minimización del error absoluto.
El STGNN_CV, en contraste, no justifica su costo computacional bajo ningún criterio
evaluado: su arquitectura de grafo, que en teoría debería aprovechar la conectividad
espacial de las cuencas, produce el mayor sobreajuste (gap = 0.109 ± 0.111), el menor
NSE (0.282) y el mayor MAPE (56.08%), lo que sugiere que la construcción del grafo de
adyacencia utilizado no captura la topología hidrológica relevante o que la densidad
de aristas es insuficiente en regiones con escasa instrumentación.

**Limitaciones y trabajo futuro.** A pesar de la solidez del marco experimental, el
estudio presenta limitaciones que condicionan la extrapolación de los resultados. La
distribución desigual de departamentos entre folds —tres en el Fold 4 (Amazonía) frente
a doce en el Fold 3 (Andes)— introduce asimetría en la estimación del error de
generalización: el fold más difícil está representado por el menor número de
departamentos, lo que incrementa la varianza del estimador de error en esa región y
puede subestimar la dificultad real de la predicción amazónica. Segundo, el análisis
de variabilidad interanual reveló que todos los modelos exhiben Pearson_r cercanas a
cero cuando se evalúa exclusivamente la componente de baja frecuencia del ciclo
hidrológico, lo que evidencia que ninguna de las arquitecturas evaluadas captura la
variabilidad interanual asociada a teleconexiones climáticas como ENSO, PDO o AMO.
La incorporación de índices climáticos globales como covariables exógenas de baja
frecuencia representaría una extensión metodológica de alto impacto potencial, ya que
las anomalías de 2024 documentadas en el análisis son consistentes con los efectos de
eventos El Niño sobre el régimen de lluvias colombiano. Tercero, todos los modelos
producen estimaciones puntuales sin cuantificación de incertidumbre, lo que limita su
utilidad en sistemas de alerta temprana donde la distribución de probabilidad del caudal
futuro —y no solo su valor esperado— es la cantidad operativamente relevante; enfoques
como el Dropout de Monte Carlo, los modelos de difusión condicional o los conjuntos
profundos (deep ensembles) podrían integrarse en futuras versiones para dotar al ECT_CV
de capacidad probabilística sin sacrificar su precisión puntual. Cuarto, el esquema de
validación cruzada anidada, aunque riguroso, evalúa la generalización sobre datos
históricos pasados; la evaluación prospectiva bajo escenarios de cambio climático
proyectados requeriría datos sintéticos o simulaciones de modelos climáticos regionales
que no fueron incorporados en este trabajo. Finalmente, la comparación estadística
mediante Bootstrap pareado se realizó sobre residuos agregados sin distinguir entre
regímenes hidrológicos (crecidas, estiajes, condiciones normales), lo que abre la
pregunta de si la superioridad del ECT_CV es uniforme a través del hidrograma completo
o si se concentra en fases particulares, una pregunta que análisis de desempeño
condicionado al régimen podrían responder en trabajos futuros.
