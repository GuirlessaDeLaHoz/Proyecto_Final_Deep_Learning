# **13. Discusión Crítica**

Los resultados obtenidos indican que la arquitectura D-DCRNN fue el modelo con mejor desempeño global entre las alternativas evaluadas. En escala logarítmica alcanzó el menor RMSE (0.406 ± 0.002), el menor MAE (0.287 ± 0.004) y el mayor coeficiente de determinación (R² = 0.804 ± 0.007). Estos resultados sugieren una capacidad superior para representar la dinámica espacio-temporal de la incidencia del dengue. Aunque el Transformer constituyó el benchmark más competitivo, presentó errores mayores (RMSE = 4.054; MAE = 2.103; R² = 0.682 en escala real) y una variabilidad entre ejecuciones considerablemente superior. La ventaja observada en D-DCRNN es consistente con su capacidad para integrar simultáneamente dependencias espaciales mediante grafos y dependencias temporales mediante mecanismos recurrentes, permitiendo modelar de forma más efectiva la propagación espacial y la evolución temporal del fenómeno epidemiológico.

Desde la perspectiva de generalización, D-DCRNN también mostró resultados favorables. La diferencia promedio entre entrenamiento y validación fue reducida (gap = 0.028 ± 0.003), manteniéndose en niveles similares a los observados para Transformer (0.025 ± 0.007) y considerablemente inferiores a los registrados por GRU, LSTM y STGNN, cuyos gaps oscilaron entre 0.31 y 0.35. Adicionalmente, D-DCRNN presentó el menor coeficiente de variación del RMSE (CV = 0.0058), evidenciando una alta estabilidad frente a distintas semillas de inicialización. En conjunto, estos resultados sugieren que el modelo no solo alcanza una elevada precisión predictiva, sino que además mantiene un comportamiento consistente y reproducible en diferentes ejecuciones.

La evidencia estadística respalda parcialmente estas diferencias. El análisis post hoc mediante Tukey HSD mostró diferencias estadísticamente significativas entre D-DCRNN y Transformer para RMSE y MAE (p < 0.001), indicando una reducción significativa de los errores de predicción. Sin embargo, para el coeficiente de determinación (R²), la diferencia entre ambos modelos no resultó estadísticamente significativa (p = 0.195), lo que sugiere que ambos poseen capacidades explicativas comparables respecto a la variabilidad de la serie. Por otra parte, D-DCRNN presentó diferencias significativas frente a GRU, LSTM y STGNN en las tres métricas evaluadas, confirmando la superioridad de su desempeño de manera consistente.

Respecto al posible trade-off entre desempeño y costo computacional, D-DCRNN incorpora una arquitectura más compleja que los modelos recurrentes tradicionales y que el propio Transformer, debido a la necesidad de procesar explícitamente relaciones espaciales mediante grafos. Esto implica mayores requerimientos computacionales durante el entrenamiento y una mayor complejidad de implementación. No obstante, la magnitud de las mejoras observadas en precisión, estabilidad y capacidad explicativa sugiere que dicho costo adicional se encuentra justificado en aplicaciones donde la calidad de la predicción constituye el criterio prioritario.

Entre las principales limitaciones del estudio se encuentra el reducido número de semillas utilizadas para la evaluación estadística, lo que limita la potencia de algunos contrastes inferenciales. Asimismo, los resultados corresponden exclusivamente al contexto epidemiológico del dengue en Colombia, por lo que la capacidad de generalización hacia otras enfermedades, regiones geográficas o escalas temporales requiere validación adicional. Adicionalmente, el test de Ljung-Box evidenció autocorrelación residual significativa, indicando que aún persisten estructuras temporales no completamente capturadas por los modelos evaluados.

Como líneas futuras de investigación, se recomienda ampliar la evaluación incorporando un mayor número de semillas y horizontes de predicción más extensos, así como explorar arquitecturas híbridas que combinen mecanismos de atención con modelos basados en grafos. También resulta relevante analizar grafos dinámicos que permitan representar cambios temporales en la conectividad entre departamentos, así como incorporar variables exógenas climáticas, demográficas y socioeconómicas que puedan contribuir a mejorar la capacidad predictiva del sistema.
