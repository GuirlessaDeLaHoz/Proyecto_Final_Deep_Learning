# **13. Discusión Crítica**

## **6.4 Discusión crítica**

Los resultados experimentales y estadísticos indican de forma consistente que el modelo **ECT presenta el mejor desempeño global** dentro del conjunto de arquitecturas evaluadas. En términos de métricas agregadas, ECT alcanzó los menores valores de error (RMSE ≈ 3.77 ± 0.18; MAE ≈ 1.94 ± 0.09) y el mayor poder explicativo (R² ≈ 0.74 ± 0.01), superando de manera estable a las arquitecturas recurrentes (LSTM, GRU), convolucionales (ConvLSTM) y al modelo STGNN. El Transformer se posiciona como el segundo modelo más competitivo, con mejor desempeño relativo que las redes recurrentes tradicionales, pero con errores sistemáticamente mayores y mayor variabilidad entre ejecuciones. En contraste, GRU, LSTM y STGNN muestran desempeños significativamente inferiores, incluyendo valores de R² cercanos a cero o negativos en algunos casos, lo que evidencia una limitada capacidad para capturar la dinámica espacio-temporal del fenómeno analizado. Esta superioridad de ECT es consistente con su capacidad de modelar simultáneamente dependencias temporales y representaciones estructurales más complejas del sistema, lo que favorece una mejor aproximación a la dinámica no lineal de la incidencia.

Desde la perspectiva de generalización, ECT también muestra el comportamiento más estable entre ejecuciones. La variabilidad entre semillas es sustancialmente menor que en el resto de modelos, lo que se refleja en desviaciones estándar reducidas en RMSE, MAE y R², así como en una mayor consistencia de los resultados bajo distintas inicializaciones. El análisis de robustez confirma este patrón, mostrando que ECT mantiene el menor nivel de dispersión en las métricas de error, mientras que modelos como LSTM, GRU y STGNN presentan alta sensibilidad a la semilla, con fluctuaciones más amplias en su desempeño. En conjunto, esto sugiere que ECT no solo minimiza el error esperado, sino que también optimiza la estabilidad del proceso de aprendizaje, lo que es un indicador relevante de generalización en entornos espacio-temporales no estacionarios.

La evidencia estadística respalda estas diferencias de manera robusta. El análisis de varianza (ANOVA) muestra diferencias significativas entre modelos en RMSE, MAE y R² (RMSE: p ≈ 6.2e-09; MAE: p ≈ 1.9e-11; R²: p ≈ 9.9e-10), mientras que el test de Friedman confirma la existencia de diferencias globales entre arquitecturas (p ≈ 0.0193). Sin embargo, los análisis post-hoc matizan estos resultados: las pruebas de Tukey HSD y los contrastes no paramétricos (Wilcoxon con corrección de Holm) indican que muchas diferencias entre modelos no son significativas a nivel pareado, lo que sugiere que las diferencias relevantes se concentran principalmente entre grupos arquitectónicos (ECT/Transformer vs. redes recurrentes y STGNN), más que entre todos los modelos de forma uniforme. En este contexto, ECT se mantiene consistentemente entre los modelos con mejor desempeño y con ventajas estadísticamente relevantes frente a las arquitecturas de menor capacidad predictiva.

En cuanto al trade-off entre desempeño y complejidad, ECT introduce una mayor sofisticación arquitectónica respecto a los modelos recurrentes tradicionales, al incorporar mecanismos que incrementan la capacidad de representación de dependencias espacio-temporales. Esto implica un mayor costo computacional durante el entrenamiento y una mayor complejidad de implementación frente a GRU y LSTM. No obstante, la magnitud de las mejoras observadas en precisión, estabilidad y capacidad explicativa sugiere que este costo adicional es justificable en escenarios donde la precisión predictiva y la robustez del modelo son criterios prioritarios. A diferencia de Transformer, que presenta mayor sensibilidad a la inicialización, ECT mantiene un comportamiento más consistente, lo que refuerza su utilidad en entornos donde la reproducibilidad es crítica.

Entre las principales limitaciones del estudio se encuentra la dependencia de un número finito de semillas y configuraciones experimentales, lo que restringe parcialmente la potencia estadística de algunos contrastes. Asimismo, el análisis se limita a un único contexto epidemiológico (incidencia de dengue en Colombia), lo que reduce la generalización externa de los resultados hacia otras enfermedades o regiones con dinámicas distintas. Adicionalmente, aunque se aplicaron pruebas de significancia global y post-hoc, la interpretación puede verse afectada por la correlación entre ejecuciones derivada de la estructura temporal de los datos, lo que podría introducir dependencia residual en algunas comparaciones.

Finalmente, como líneas futuras de investigación se recomienda ampliar el análisis mediante validación espacio-temporal más estricta, incrementar el número de semillas y horizontes de predicción, e incorporar esquemas de validación externa en otros contextos epidemiológicos. También resulta pertinente explorar arquitecturas híbridas que integren mecanismos de atención con grafos dinámicos, así como la incorporación de variables exógenas climáticas, demográficas y socioeconómicas que permitan enriquecer la representación del sistema y potencialmente mejorar la capacidad predictiva global del modelo.
